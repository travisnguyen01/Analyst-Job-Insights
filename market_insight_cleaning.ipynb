```python
from google.colab import files
uploaded = files.upload()
import pandas as pd

df = pd.read_excel("data_analysis_role_scraping.xlsx")
df = df.dropna(subset=["Job Title"])
df["Salary Type"] = df["Salary Type"].fillna("unspecified")
df["Salary Type"].value_counts()
df["Work Type"] = df["Work Type"].str.lower().str.strip()
df["Work Type"].value_counts()
df["Role Level"] = df["Role Level"].str.lower().str.strip()
df["Role Level"].value_counts()
skills_list = ["sql", "excel", "python", "tableau", "power bi"]
def extract_skills(text):
    text = str(text).lower()
    found = [skill for skill in skills_list if skill in text]
    return ", ".join(found) if found else "unspecified"
df["Skills"] = df["Description"].apply(extract_skills)
df["Skills"].value_counts()

df.info()
df.head(27)
```

```python
df["Role Level"].value_counts()
```

```python
from collections import Counter

all_skills = []

for skills in df["Skills"]:
    if skills != "unspecified":
        split_skills = skills.split(", ")
        all_skills.extend(split_skills)

skill_counts = Counter(all_skills)

skill_counts
```

```python
skill_df = pd.DataFrame(skill_counts.items(), columns=["Skill", "Count"])
skill_df = skill_df.sort_values(by="Count", ascending=False)

skill_df
```
